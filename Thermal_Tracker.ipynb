```python
import os
import cv2

train_dataset_path = "/content/drive/MyDrive/CV_datasets/Thermal_dataset/Dataset/train"
val_dataset_path = "/content/drive/MyDrive/CV_datasets/Thermal_dataset/Dataset/val"

count_512x640 = 0
count_512x512 = 0
total = 0

def process_folder(path):
    global count_512x640, count_512x512, total

    for root, dirs, files in os.walk(path):  # 🔥 bütün subfolderləri gəzir
        for file in files:
            if file.lower().endswith((".jpg", ".png", ".jpeg")):

                img_path = os.path.join(root, file)
                img = cv2.imread(img_path)

                if img is None:
                    continue

                h, w = img.shape[:2]
                total += 1

                if h == 512 and w == 640:
                    count_512x640 += 1
                elif h == 512 and w == 512:
                    count_512x512 += 1

# train + val
process_folder(train_dataset_path)
process_folder(val_dataset_path)

print("Total images:", total)
print("512x640 images:", count_512x640)
print("512x512 images:", count_512x512)
```
